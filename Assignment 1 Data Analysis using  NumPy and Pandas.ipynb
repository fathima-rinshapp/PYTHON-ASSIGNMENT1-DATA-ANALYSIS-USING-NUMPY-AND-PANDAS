# pandas

```python
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
```

```python
marks=pd.Series(data=[95, 92, 89, 85, 80],index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'],name='mark')
print(marks)
```

```python
print('mark of the 1st rank student=',marks[0])
```

```python
#marks of the top 3 ranks by specifying their index labels.
print(marks.loc['Rank1':'Rank3'])
```

```python
#Use the iloc accessor to retrieve the mark of the 3rd rank student.
print(marks.iloc[2])
```

```python
#a boolean mask to filter the Series and retrieve the ranks where the marks are greater than 90.
print(marks[marks>90])
```

```python
#Modify the mark of the student with the 1st rank to 100.
marks['Rank1']=100
print(marks)
```

```python
#Remove the entry corresponding to the last rank from the Series.
marks=marks.drop('Rank5')
print(marks)
```

```python
#Compute the CGPA by dividing each mark in the Series by 10.
CGPA=marks/10
print(CGPA)
```

## DataFrame

```python
info={'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
      'ProductCategory':['Electronics', 'Clothing', 'Electronics', 'Furniture',
'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
'Region':['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West',
'South', 'North'],
'Amount':[200, 150, 300, 450, 200, 250, 300, 180, 350, 400]}
transactions=pd.DataFrame(info)
```

```python
print(transactions)
print()
print(transactions.head())
print()
print(transactions.tail())
print()
print('shape= ',transactions.shape)
print()
print('columns=',
      transactions.columns)
print()
print('datatype',
      transactions.dtypes)
```

```python
print(transactions[['ProductCategory','Amount']])
```

```python
#Retrieve the last 3 columns of the DataFrame using iloc.
print(transactions.iloc[:,-3:])
```

```python
#Filter rows where the 'Region' is 'North' and 'Amount' is greater than 200.
transactions[(transactions['Region']=='North')&(transactions['Amount']>200)]
```

```python
#the value counts for the 'ProductCategory' column.
transactions['ProductCategory'].value_counts()
```

```python
#Find the unique values in the 'Region' column.
transactions['Region'].unique()
```

```python
#Group by 'Region' and find the mean amount for each region.
transactions.groupby('Region')['Amount'].mean()
```

```python
#Modify the 'Amount' for the transaction with TransactionID 102 to 165.
transactions.loc[transactions['TransactionID']==102,'Amount']=165
print(transactions)
```

```python
#adding new columns
transactions['Discount']=transactions['Amount']*0.10
print(transactions)
```

```python
#Remove the row with TransactionID 109.
transactions=transactions.drop(transactions[transactions['TransactionID']==109].index)
print(transactions)
```

```python
transactions.drop('Discount',axis=1,inplace=True)
print(transactions)
```
